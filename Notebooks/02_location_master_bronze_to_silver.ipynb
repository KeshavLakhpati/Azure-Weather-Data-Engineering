```python
from pyspark.sql.types import StructType
import pyspark.sql.functions as F


flattened_df = df

# Loop through all columns in the dataframe
for col_name, col_type in df.dtypes:
    # Check if the column is a Struct (parsed JSON)
    if col_type.startswith("struct"):
        # Get all sub-fields of this struct
        struct_fields = [field.name for field in df.schema[col_name].dataType]
        
        # Add each sub-field as a new column with a prefix to avoid name collisions
        for field in struct_fields:
            flattened_df = flattened_df.withColumn(
                f"{col_name}_{field}", 
                F.col(f"{col_name}.{field}")
            )
        
        # Optionally, drop the original nested struct column
        flattened_df = flattened_df.drop(col_name)

#display(flattened_df)
```

```python
from pyspark.sql.types import StructType

def flatten_dataframe(df):
    complex_fields = dict([(field.name, field.dataType) 
                           for field in df.schema 
                           if isinstance(field.dataType, StructType)])
    
    while len(complex_fields) > 0:
        col_name = list(complex_fields.keys())[0]
        #print(f"Flattening nested structure: {col_name}")
        
        # Get sub-fields
        sub_fields = [field.name for field in complex_fields[col_name].fields]
        
        # Create select expression
        select_expr = [F.col(f"`{col_name}`.`{field}`").alias(f"{col_name}_{field}") for field in sub_fields]
        
        # Add non-nested columns to the expression so they aren't lost
        other_cols = [F.col(f"`{c}`") for c in df.columns if c != col_name]
        
        df = df.select(*other_cols, *select_expr)
        
        # Re-evaluate remaining complex fields
        complex_fields = dict([(field.name, field.dataType) 
                               for field in df.schema 
                               if isinstance(field.dataType, StructType)])
    return df

# Run the function on your dataframe
fully_flattened_df = flatten_dataframe(df)
#display(fully_flattened_df)
```

```python
silver_df = fully_flattened_df.select("EnglishName","Key","GeoPosition_Latitude","GeoPosition_Longitude").where(fully_flattened_df.AdministrativeArea_ID.isin(['MH','DL']))
```

```python
silver_path = "abfss://silver@stweatherprojectnew.dfs.core.windows.net/location_master/"
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(silver_path)
```
