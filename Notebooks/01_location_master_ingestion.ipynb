```python
import json
import requests
from datetime import datetime

base_path = "bronze-container-path"
api_key = "api-key"
cities = [
    "Pune",
    "Mumbai",
    "Delhi"
]

for city in cities:

    url = f"https://dataservice.accuweather.com/locations/v1/cities/search?apikey={api_key}&q={city}"

    response = requests.get(url)

    data = response.json()

    json_data = json.dumps(data, indent=4)

    file_name = f"{city.lower()}_{datetime.now().strftime('%Y%m%d')}.json"

    dbutils.fs.put(base_path + file_name, json_data, True)

    print(f"Saved data for {city}")
```
