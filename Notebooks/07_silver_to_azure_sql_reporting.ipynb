Finally, connect your Silver table directly into your reporting layer. It skips right to your target Azure SQL database destination.

```python
# Databricks notebook source
# 1. Database Connection String Configs
jdbc_hostname = "weather-sql-server-new.database.windows.net" 
jdbc_port     = "1433"
jdbc_database = "db_name"
jdbc_username = ""
#jdbc_password = dbutils.secrets.get(scope="weather-scope", key="sql-password") 
jdbc_password = ""

jdbc_url = f"jdbc:sqlserver://{jdbc_hostname}:{jdbc_port};database={jdbc_database};encrypt=true;trustServerCertificate=false;"

# 2. Read directly from the processed Silver Delta table
silver_df = spark.table("weather_catalog.silver.forecasts")

# 3. Sync to Azure SQL Reporting
print("Syncing cleaned data from Silver to Azure SQL database reporting table...")
silver_df.write \
    .format("jdbc") \
    .option("url", jdbc_url) \
    .option("dbtable", "dbo.forecasts_reporting") \
    .option("user", jdbc_username) \
    .option("password", jdbc_password) \
    .mode("overwrite") \
    .save()

print("Medallion to Database flow completed successfully!")
```
